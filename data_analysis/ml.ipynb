#### Outliers

```python
#Practice for Outliers

np.random.seed(100)
data = np.random.normal(size=10000) #<=== normal?????
data = pd.Series(data)  #<=== Series vs dataFrame ?????
data.describe()
```

```python
min, q1, q2, q3, max = np.percentile(data, [0, 25, 50, 75, 100])
min, q1, q2, q3, max
IQR = q3 - q1 #<=== InterQuartile
lower_limit = q1 - 1.5 * IQR
upper_limit = q3 + 1.5 * IQR
lower_limit, upper_limit
```

```python
data = np.where(data < lower_limit, lower_limit, data)
data = np.where(data > upper_limit, upper_limit, data)
plt.boxplot(data, vert=False);
# So outliers are extreme values has impact on data
# inter quartile (IQR) and boxplots are useful in identifying the outliers
# whiskers show the upper and lower limits
# values outside the limits can be eliminated before analysis 
```

```python
#Manually correlation calculation

cor_climb_dist = (((hills.climb - hills.climb.mean()) * (hills.dist - hills.dist.mean())).sum()/34)/(hills.climb.std()*hills.dist.std())
cor_climb_dist
```

```python
cor_dist_time = ((hills.dist - hills.dist.mean()) * (hills.time - hills.time.mean())).sum()/(len(hills)-1)/(hills.dist.std() * hills.time.std())
cor_dist_time
```

```python
cor_climb_time = ((hills.climb - hills.climb.mean())*(hills.time - hills.time.mean())).sum()/(len(hills)-1)/(hills.climb.std()*hills.time.std())
cor_climb_time
```
