> Bivariate analysis

- to understand relation between to variables
- to study the effect of one variable (independent) on the other (dependent) variable 


```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

hills = pd.read_csv('../data_analysis/sources/Hills.csv')
hills.head(5)
```

```python
hills.describe()
```

```python
sns.distplot(hills.dist)
```

```python
#Visual with scatterplot
sns.scatterplot(x= hills.dist, y= hills.time);

#numarical (independent variable = dist) and numerical (dependent variable = climb)
```

```python
sns.scatterplot(x=hills.climb, y=hills.time)
```

```python
sns.scatterplot(x=hills.dist, y=hills.climb);
```

```python
#Pairplot
sns.pairplot(hills);
```

```python
sns.jointplot(x=hills.dist, y=hills.time, kind='kde');
```

```python
sns.jointplot(x=hills.dist, y=hills.time, kind='reg');
plt.hist(hills.dist) # ===> histogram
```

```python
# Covarince n-1 for sample, for population just n =====> covariance is bivariate as you see, x and y
cov_hill = hills.cov()
print(cov_hill)
cov_dist_time = ((hills.dist - hills.dist.mean()) * (hills.time - hills.time.mean())).sum() / (len(hills) - 1)
print(cov_dist_time)
cov_climb_time = ((hills.climb - hills.climb.mean()) * (hills.time - hills.time.mean())).sum() / (len(hills) - 1)
print(cov_climb_time)
```

```python
# correlation - =====> covariance is bivariate as you see, x and y

corr_climb_time = cov_climb_time / (hills.climb.std() * hills.time.std())
corr_dist_time = cov_dist_time / (hills.dist.std() * hills.time.std())
print(corr_climb_time)
print(corr_dist_time)
hills.corr()
```
