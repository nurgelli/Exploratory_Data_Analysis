# Create normal distribution population

```python
import seaborn as sn
import random
import pandas as pd
import numpy as np  

random.seed(1)
norm_pop = np.random.normal(10, 2, size=1000)
sn.kdeplot(norm_pop)
```

### Create a sampling distribution by sampling the normal distribution population

```python
def sampling_distr(pop, n):
    sample = []
    for i in np.arange(0, 100, 1):
        np.random.seed(1)  # to fix randomness
        sample.append(np.array(random.sample(list(pop), n)).mean()) # if sampling_distr(norm_pop, 10) ==> then 10 sample from each iteration and mean of those 10 sample 
    return sample
```

```python
sn.kdeplot(norm_pop)
sn.kdeplot(sampling_distr(norm_pop, 10)); # 10 is means of 10 values
sn.kdeplot(sampling_distr(norm_pop, 50)); # means of 50 values
```

```python
norm_pop.mean(), np.array(sampling_distr(norm_pop, 10)).mean(), np.array(sampling_distr(norm_pop, 10)).mean() # all means are almost same
```

```python
norm_pop.std(), np.array(sampling_distr(norm_pop, 10)).std(), np.array(sampling_distr(norm_pop, 50)).std()
```

```python
2/(10 ** 0.5) #==> 2 is std of population, 10 sample size === this is Standard Error  = sigma / sqrt(n) ===> sigma by **0.5 is sqrt of n-sample_size
```

```python
2/(50 ** 0.5) #==> 2 is std of population, 10 sample size === this is Standard Error
```

## Creating exponential distribution

### Create a sampling distribution by sampling the exponential distribution population

```python
sn.kdeplot(exp_pop)
sn.kdeplot(sampling_distr(exp_pop, 10))
sn.kdeplot(sampling_distr(exp_pop, 50))
```
