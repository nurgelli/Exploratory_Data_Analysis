```python
import scipy.stats as ss
import numpy as np
import pandas as pd
import seaborn as sns


sns.kdeplot(ss.norm.rvs(loc=60, scale=12, size=1000, random_state=4))
```

```python
# cut-off marks for the top 5
ss.norm.ppf(q=0.95, loc=60, scale=12)

# ppf ==> percent point function reverse of cdf  
```

```python
# cut-off marks for the top 10
ss.norm.ppf(q=0.9, loc=60, scale=12)
```

```python
# bottom 10
ss.norm.ppf(q=0.1, loc=60, scale=12)
```
