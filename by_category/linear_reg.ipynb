# Goodness of Fit


```python
import pandas as pd 
import numpy as np
import seaborn as sn
from scipy import stats
import matplotlib.pyplot as plt

plt.rcParams['figure.figsize'] = (15, 6)
```

```python
df = pd.read_csv('../data_analysis/sources/M6_T1_V4_BrakingDistance.csv')
df.head(5)
```

```python
df.info()
```

```python
df.describe()
```

```python
plt.scatter(df.speed, df.dist)
plt.title("Speed vs Distance")
plt.xlabel("Speed")
plt.ylabel("Distance")
# Positive linear regression
```

```python
df.corr()

# which is R => 0.806
```

```python
sn.displot(df.speed)
```

```python
xbar = df.speed.mean()
xbar
```

```python
ybar = df.dist.mean()
ybar
```

```python
beta1 = np.sum((df.dist - ybar) * (df.speed - xbar)) / np.sum((df.speed - xbar)**2)
beta1 # <=== so beta1 is 3.93
```

```python
beta0 = ybar - beta1 * xbar
beta0
```
