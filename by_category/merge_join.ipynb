#### MERGE
>Bu konu SQL'deki ***JOIN***'in pandas karşılığı — ve ikisini birlikte öğrenmek çok mantıklı çünkü mülakatlarda ikisi de soruluyor.

```python
import pandas as pd

orders = pd.DataFrame({
    "order_id": [1001, 1002, 1003],
    "customer_id": [42, 55, 46]
})

customers = pd.DataFrame({
    "customer_id": [42, 55],
    "name": ["Dave", "Nancy"],
    "city": ["Ankara", "Istanbul"]
})

employees = pd.DataFrame({
    "id": [42, 55],
    "employee": ["Ali", "Jordan"],
    
})
```

```python
# inner join
pd.merge(orders, customers, on="customer_id", how='inner')
```

```python
pd.merge(orders, customers, on="customer_id", how="left") # <=== as seen orders table is in the left, so got left table
# and just matched one from right table (in my case customers)

# left table olup right tablede olmayanlar NaN olarak gelir, cunku left baz alindi, right tam tersi
```

```python
pd.merge(orders, customers, on="customer_id", how='right') # <== right table is customer and just got what matched from 
# left table
```

```python
pd.merge(orders, customers, on="customer_id", how='outer') # what matched was got from both of tables
```

```python
new_table = pd.merge(employees, customers, left_on="id", right_on='customer_id', how='left') #outer, right, left, inner
```

```python
pd.merge(new_table, orders, on='customer_id', how='right')
```

> left join → sol tablonun satır sayısı hiç değişmez. 3 satır girdiyse 3 satır çıkar.
